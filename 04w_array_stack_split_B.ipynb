## 4.1 배열 결합

- `vstack()` : 세로 방향 결합
- `hstack()` : 가로 방향 결합
- `column_stack()` : 열 방향 결합
- `row_stack()` : `vstack()`과 거의 같은 기능
- `concatenate()` : 축을 지정해서 결합
- `stack()` : 새 축을 추가해서 차원을 하나 늘려 결합

### 1. `np.vstack()` : 2차원 배열 세로 결합

- 축 0 방향으로 결합
- 열 수가 같아야 함

```python
# np.vstack()
# 3행 2열 배열 생성
a = np.arange(6).reshape(3, 2)
a
```

```python
# 같은 열 수를 가진 배열 생성
b = np.arange(6, 12).reshape(3, 2)
b
```

```python
# 세로로 결합
# 결과: 6행 2열
np.vstack((a, b))
```

### 2. `np.vstack()` : 1차원 배열 결합

- 1차원 배열을 세로로 쌓으면 2차원 배열이 됨
- 각 배열의 길이가 같아야 함

```python
# 1차원 배열 2개 생성
x = np.array([1, 2, 3])
y = np.array([5, 6, 7])

# 1D -> 2D
np.vstack((x, y))
```

### 3. `np.hstack()` : 2차원 배열 가로 결합

- 축 1 방향으로 결합
- 행 수가 같아야 함

```python
# 3행 2열 배열 생성
a = np.arange(6).reshape(3, 2)
a
```

```python
# 3행 3열 배열 생성
b = np.arange(10, 19).reshape(3, 3)
b
```

```python
# 가로로 결합
# 결과: 3행 5열
np.hstack((a, b))
```

### 4. `np.hstack()` : 1차원 배열 결합

- 1차원 배열은 길이가 달라도 이어 붙일 수 있음
- 결과도 1차원 배열

```python
# 1차원 배열 수평 결합
np.hstack((np.arange(3), np.arange(10, 15)))
```

### 5. `np.column_stack()` : 1차원 배열을 열로 결합

- 1차원 배열을 자동으로 열 벡터처럼 바꾼 후 결합
- `hstack()`과 결과가 다를 수 있음

```python
# 1차원 배열 생성
x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

# 열 방향으로 결합
np.column_stack((x, y))
```

```python
# 1차원 배열 x의 모양 확인
x.shape
```

```python
# np.newaxis로 열 방향 차원 추가
# (3,) -> (3, 1)
x[:, np.newaxis]
```

```python
# None도 np.newaxis와 같은 역할
x[:, None]
```

```python
# 이미 열 벡터로 바꾼 뒤 column_stack 적용
np.column_stack((x[:, np.newaxis], y[:, np.newaxis]))
```

```python
# hstack은 1차원 배열을 그대로 이어 붙임
np.hstack((x, y))
```

```python
# column_stack과 hstack은 같은 함수가 아님
np.column_stack == np.hstack
```

### 6. `column_stack()`과 `hstack()`의 차이 보충

- 1차원 배열에서는 결과가 다름
- 2차원 배열에서는 비슷하게 동작할 수 있음

```python
# 1차원 배열 준비
x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

print('column_stack 결과')
print(np.column_stack((x, y)))

print('hstack 결과')
print(np.hstack((x, y)))
```

```python
# 2차원 배열 준비
x2 = np.arange(1, 7).reshape(2, 3)
y2 = np.arange(7, 13).reshape(2, 3)

print(x2)
print(y2)
```

```python
# 2차원에서는 둘 다 가로 결합처럼 동작
print(np.column_stack((x2, y2)))
print(np.hstack((x2, y2)))
```

### 8. `np.concatenate()` : 1차원 배열 결합

- 기본은 `axis=0`
- 1차원 배열은 축 0만 존재
- `axis=1`은 오류 발생

```python
# 1차원 배열 3개 생성
a = np.array([1, 2])
b = np.array([3, 4])
c = np.array([5, 6])

a, b, c
```

```python
# 기본 결합
np.concatenate((a, b, c))
```

```python
# axis=0 결합
np.concatenate((a, b, c), axis=0)
```

```python
# axis=None이면 모두 펼쳐서 결합
np.concatenate((a, b, c), axis=None)
```

```python
# 오류 예시
# 1차원 배열에는 axis=1이 없음
np.concatenate((a, b, c), axis=1)
```

### 9. `np.concatenate()` : 2차원 배열 결합

- `axis=0` : 세로 결합, 열 수가 같아야 함
- `axis=1` : 가로 결합, 행 수가 같아야 함
- `axis=None` : 배열을 1차원으로 펼쳐 결합

```python
# 2행 3열 배열
x = np.arange(6).reshape(2, 3)
x
```

```python
# 2행 3열 배열
y = np.arange(10, 16).reshape(2, 3)
y
```

```python
# 2행 2열 배열
z = np.arange(20, 24).reshape(2, 2)
z
```

```python
# 기본 axis=0
# 세로 결합
np.concatenate((x, y))
```

```python
# axis=0 명시
np.concatenate((x, y), axis=0)
```

```python
# vstack과 같은 결과
np.vstack((x, y))
```

```python
# axis=1
# 가로 결합
np.concatenate((x, y), axis=1)
```

```python
# x와 z는 행 수가 같으므로 axis=1 가능
np.concatenate((x, z), axis=1)
```

```python
# axis=None은 평면화 후 연결
np.concatenate((x, y), axis=None)
```

```python
# 오류 예시
# axis=0은 열 수가 같아야 하는데 x와 z는 열 수가 다름
np.concatenate((x, z), axis=0)
```

### 13. `np.stack(axis=2)` 과정 확인

- `np.newaxis` 또는 `expand_dims()`로 축 2를 추가
- 그 뒤 `concatenate(axis=2)`로 붙이는 것과 비슷하게 이해 가능

```python
# x에 마지막 축 추가
x[:, :, np.newaxis]
```

```python
# y에 마지막 축 추가
y[:, :, np.newaxis]
```

```python
# z에 마지막 축 추가
z[:, :, np.newaxis]
```

```python
# expand_dims로 같은 결과 확인
np.expand_dims(x, axis=2)
```

```python
# axis=2로 확장한 배열을 concatenate
np.concatenate((x[:, :, np.newaxis], y[:, :, np.newaxis], z[:, :, np.newaxis]), axis=2)
```

```python
# stack(axis=2)와 같은 결과인지 확인
np.array_equal(
    np.stack((x, y, z), axis=2),
    np.concatenate((x[:, :, np.newaxis], y[:, :, np.newaxis], z[:, :, np.newaxis]), axis=2)
)
```

### 14. 열 벡터로 바꾼 뒤 결합하기

- `hstack()`은 2차원 배열끼리 결합해야 안정적
- 1차원 배열 `c`는 `(3, 1)`로 바꿔 붙임

```python
# 3행 4열 배열과 1차원 배열 생성
a = np.arange(1, 13).reshape(3, 4)
c = np.arange(1, 6, 2)
a, c
```

```python
# expand_dims 사용
cc = np.expand_dims(c, axis=1)
cc
```

```python
# 오류 예시
# 2차원 배열 a와 1차원 배열 c는 hstack에서 차원이 맞지 않음
np.hstack((a, c))
```

### 21. `np.r_['i, j, k', ...]` : 차원 확장 위치 지정

```python
# 결과 모양 확인
np.r_['0,3,0', a, b].shape
```

```python
# 최소 3차원, k=1
# (3,) -> (1, 3, 1)
np.r_['0,3,1', a, b]
```

```python
# 결과 모양 확인
np.r_['0,3,1', a, b].shape
```

```python
# 최소 3차원, k=2
# (3,) -> (1, 1, 3)
np.r_['0,3,2', a, b]
```

```python
# 결과 모양 확인
np.r_['0,3,2', a, b].shape
```

### 23. 1차원 배열 결합 방법 정리 코드

- 수평 결합: `hstack`, `r_`, `concatenate`
- 수직 결합: `vstack`, `row_stack`, `r_['0,2']`
- 열 결합: `column_stack`, `c_`, `r_['-1,2,0']`

```python
# 1차원 배열 준비
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print('수평 결합')
print(np.hstack((a, b)))
print(np.r_[a, b])
print(np.concatenate((a, b)))

print()
print('수직 결합')
print(np.vstack((a, b)))
print(np.row_stack((a, b)))
print(np.r_['0,2', a, b])

print()
print('열 결합')
print(np.column_stack((a, b)))
print(np.c_[a, b])
print(np.r_['-1,2,0', a, b])
```

## 4.2 배열 분리

- `split()` : 배열을 여러 부분으로 분리
- `hsplit()` : 축 1 방향 분리
- `vsplit()` : 축 0 방향 분리
- `array_split()` : 균등하게 나누어지지 않아도 분리 가능
- `dsplit()` : 3차원 이상에서 축 2 방향 분리

### 1. `np.split()` : 1차원 배열 균등 분리

- 반환값은 배열들의 리스트
- 정수로 나눌 때는 같은 크기로 나누어져야 함

```python
# 0.0부터 8.0까지 1차원 배열
x = np.arange(9.0)
x
```

```python
# 3개로 균등 분리
s = np.split(x, 3)
s
```

```python
# 분리 결과 자료형 확인
print(type(s))
for part in s:
    print(part, part.shape)
```

### 2. `np.split()` : 인덱스 위치 기준 분리

- `[i, j, k]`를 넣으면 구간 기준으로 분리
- 예: `[2, 6]`이면 `[:2]`, `[2:6]`, `[6:]`

```python
# 10부터 60까지 10 간격 배열
y = np.arange(10, 61, 10)
y
```

```python
# 인덱스 기준으로 여러 구간 분리
np.split(y, [2, 3, 4, 6])
```

```python
# 1부터 9까지 배열
x = np.arange(1, 10)
x
```

```python
# [2, 6] 기준 분리
np.split(x, [2, 6])
```

### 3. 2차원 배열의 `split(axis=1)`과 `hsplit()`

- 축 1 방향으로 분리
- 열 방향으로 자른다고 이해

```python
# 2행 12열 배열 생성
a = np.arange(24).reshape(2, 12)
a
```

```python
# hsplit으로 3개 분리
np.hsplit(a, 3)
```

```python
# split(axis=1)도 같은 기능
np.split(a, 3, axis=1)
```

```python
# 4개로도 균등 분리 가능
np.split(a, 4, axis=1)
```

```python
# 인덱스 기준으로 열 분리
np.split(a, [3, 5, 7], axis=1)
```

### 4. 2차원 배열의 `split(axis=0)`과 `vsplit()`

- 축 0 방향으로 분리
- 행 방향으로 자른다고 이해

```python
# 6행 4열 배열
b = np.arange(24).reshape(6, 4)
b
```

```python
# 축 0 기준으로 3개 분리
np.split(b, 3, axis=0)
```

```python
# 인덱스 기준 분리
np.split(b, [2, 4, 6], axis=0)
```

### 5. `np.hsplit()` : 1차원 배열과 2차원 배열

- 1차원 배열도 분리 가능
- 2차원 배열에서는 열 방향 분리

```python
# 1차원 배열
arr = np.arange(10)
arr
```

```python
# 2개로 균등 분리
np.hsplit(arr, 2)
```

```python
# 인덱스 기준 분리
np.hsplit(arr, [2, 5])
```

```python
# 2차원 배열 준비
a = np.arange(24).reshape(2, 12)
a
```

```python
# (3, 5, 8) 기준으로 열 분리
np.hsplit(a, (3, 5, 8))
```

```python
# 앞부분 슬라이싱 확인
a[:, :3]
```

```python
# 마지막 부분 슬라이싱 확인
a[:, 8:]
```

### 6. `np.vsplit()` : 2차원 배열 행 방향 분리

- 축 0 기준으로 분리
- 균등 분리가 안 되면 오류 발생

```python
# 6행 3열 배열
a = np.arange(18).reshape(6, 3)
a
```

```python
# 3개로 균등 분리
np.vsplit(a, 3)
```

```python
# 오류 예시
# 6행을 4개로 균등 분리할 수 없음
np.vsplit(a, 4)
```

```python
# 인덱스 기준 분리
np.vsplit(a, (1, 4, 6))
```

```python
# 슬라이싱으로 같은 구간 확인
a[:1, :]
```

```python
# 중간 구간 확인
a[4:6, :]
```

```python
# 마지막 빈 구간 확인
a[6:, :]
```

### 7. 3차원 배열의 `vsplit()`

- `vsplit()`은 무조건 축 0 기준
- `np.split(x, n, axis=0)`과 같음

```python
# 4행 4열 2개 깊이의 3차원 배열
x = np.arange(32).reshape(4, 4, 2)
x
```

```python
# split(axis=0)
np.split(x, 2, axis=0)
```

```python
# vsplit과 같은 결과
np.vsplit(x, 2)
```

```python
# 축 0 기준 슬라이싱
x[:1, :, :]
```

```python
# 축 0 기준 중간 구간
x[1:3]
```

```python
# 축 0 기준 마지막 구간
x[3:]
```

```python
# 인덱스 [1, 3] 기준 분리
np.vsplit(x, [1, 3])
```

### 8. 3차원 배열의 `hsplit()`

- 3차원 이상에서 `hsplit()`은 축 1 기준
- `np.split(x, n, axis=1)`과 같음

```python
# 3차원 배열 확인
x = np.arange(32).reshape(4, 4, 2)
x
```

```python
# hsplit으로 2개 분리
np.hsplit(x, 2)
```

```python
# split(axis=1)과 같은 결과
np.split(x, 2, axis=1)
```

```python
# 인덱스 기준 분리
np.hsplit(x, [1, 3])
```

### 9. `np.array_split()` : 균등하지 않아도 분리 가능

- `split()`은 균등하게 안 나누어지면 오류
- `array_split()`은 가능한 크기로 나누어 반환

```python
# 길이 9인 배열
x = np.arange(9)
x
```

```python
# 5개로 균등하지 않게 분리 가능
np.array_split(x, 5)
```

```python
# split은 같은 상황에서 오류 발생
np.split(x, 5)
```

### 10. 2차원 `np.array_split()`

- `axis=0` : 행 방향 분리
- `axis=1` : 열 방향 분리
- 인덱스 리스트도 사용 가능

```python
# 4행 6열 배열
x = np.arange(24).reshape(4, 6)
x
```

```python
# axis=0 기준 3개 분리
np.array_split(x, 3, axis=0)
```

```python
# axis=1 기준 4개 분리
np.array_split(x, 4, axis=1)
```

```python
# [2, 4] 기준 axis=0 분리
np.array_split(x, [2, 4], axis=0)
```

```python
# [2, 4] 기준 axis=1 분리
np.array_split(x, [2, 4], axis=1)
```

### 11. 3차원 배열에서 `array_split(axis=2)`

- 3차원 배열은 축 0, 1, 2를 가짐
- `axis=2`는 마지막 축 방향 분리

```python
# shape (2, 2, 6)
a = np.arange(24).reshape(2, 2, 6)
a
```

```python
# 마지막 축 기준으로 4개 분리
np.array_split(a, 4, axis=2)
```

```python
# 각 부분 배열의 shape 확인
for part in np.array_split(a, 4, axis=2):
    print(part.shape)
```

```python
# shape (3, 3, 4) 예시
b = np.arange(36).reshape(3, 3, 4)
b
```

```python
# axis=2 기준 2개 분리
np.array_split(b, 2, axis=2)
```

### 12. `np.dsplit()` : 3차원 이상에서 깊이 방향 분리

- `dsplit()`은 축 2 기준
- 3차원 이상의 배열에만 사용 가능

```python
# 3차원 배열 생성
x = np.arange(24).reshape(2, 2, 6)
x
```

```python
# dsplit으로 3개 분리
np.dsplit(x, 3)
```

```python
# 인덱스 기준 분리
np.dsplit(x, [2, 4, 5])
```

```python
# 오류 예시
# 6개를 4개로 균등 분리할 수 없음
np.dsplit(x, 4)
```

```python
# 오류 예시
# dsplit은 3차원 이상에서 사용 가능
np.dsplit(np.arange(6).reshape(2, 3), 1)
```
