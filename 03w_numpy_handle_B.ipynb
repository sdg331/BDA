## 3.1 패키지 numpy 개요와 자료형 ndarray

- numpy는 수치 계산을 빠르게 처리하는 패키지
- 핵심 자료형은 `ndarray`
- `ndarray`는 같은 자료형의 원소로 구성됨
- 배열의 차원, 모양, 자료형 등을 속성으로 확인 가능

```python
# numpy 불러오기
# np라는 별명으로 짧게 사용
import numpy as np
```

### ndarray 생성

- `np.array()` : 리스트나 튜플을 배열로 변환
- `np.ndarray()` : 지정한 모양의 배열 생성
- `np.ndarray()`는 값을 직접 넣지 않으면 기존 메모리 값이 들어갈 수 있음

```python
# 튜플을 ndarray로 변환
a = np.array((2, 3))
a
```

```python
# ndarray 직접 생성
# 값은 초기화되지 않아서 실행 환경마다 다르게 보일 수 있음
a = np.ndarray((3, 2), dtype=int)
a
```

```python
# dtype을 두 번째 인자로 지정
a = np.ndarray((3, 2), int)
a
```

### array()와 주요 속성 확인

- `shape` : 배열의 모양
- `ndim` : 차원 수
- `dtype` : 원소 자료형
- `size` : 전체 원소 개수
- `itemsize` : 원소 1개의 바이트 크기
- `nbytes` : 배열 전체 바이트 크기

```python
# int 자료형 배열 생성
a = np.array([1, 2, 3], dtype=int)
a
```

```python
# 배열 객체의 자료형 확인
type(a)
```

```python
# 배열의 기본 속성 확인
print(a.shape)
print(a.ndim)
print(a.dtype)

# 실행 환경에 따라 int32, int64처럼 다르게 나올 수 있음
```

```python
# 원소 개수와 메모리 크기 확인
print(a.size)
print(a.itemsize)
print(a.nbytes)
```

## 3.2 내장함수 arange()와 linspace()

- `np.arange()` : 시작, 끝, 간격을 기준으로 배열 생성
- `np.linspace()` : 시작, 끝, 개수를 기준으로 배열 생성
- `reshape()` : 원소 개수는 유지하면서 배열 모양 변경

### np.arange()

- `np.arange(stop)`
- `np.arange(start, stop)`
- `np.arange(start, stop, step)`
- stop 값은 포함되지 않음

```python
# 정수 범위 배열 생성
print(np.arange(3))
print(np.arange(3, 7))
print(np.arange(3, 7, 2))
```

```python
# 실수 범위도 사용 가능
print(np.arange(2.0))
print(np.arange(2.0, 6.5))
print(np.arange(2.0, 6.5, 0.8))
```

### reshape()

- 배열의 모양을 바꿈
- 전체 원소 개수는 같아야 함
- 원소 개수가 맞지 않으면 오류 발생

```python
# 0부터 11까지 12개 원소 생성 후 3행 4열로 변경
a = np.arange(12).reshape(3, 4)
a
```

```python
# 12개 원소를 2행 6열로 변경
np.reshape(a, (2, 6))
```

```python
# 오류 예시
# 12개 원소를 3행 5열로 바꾸려고 해서 오류 발생
np.reshape(a, (3, 5))
```

### np.linspace()

- `start` : 시작값
- `stop` : 끝값
- `num` : 생성할 값의 개수
- 기본적으로 끝값을 포함함
- 간격은 자동 계산

```python
# 2.0부터 3.0까지 기본 50개 생성
np.linspace(2.0, 3.0)
```

```python
# retstep=True를 사용하면 간격도 같이 반환
np.linspace(2.0, 3.0, retstep=True)
```

```python
# 기본 50개일 때 간격 계산
print((3 - 2) / (50 - 1))
```

```python
# endpoint=False면 stop 값 제외
np.linspace(2.0, 3.0, endpoint=False, retstep=True)
```

```python
# endpoint=False일 때 간격 계산
print((3 - 2) / 50)
```

```python
# 5개의 값 생성
np.linspace(2.0, 3.0, num=5, retstep=True)
```

```python
# 5개 생성 + 끝값 제외
np.linspace(2.0, 3.0, num=5, endpoint=False, retstep=True)
```

### arange와 linspace 비교

- `arange`는 간격 중심
- `linspace`는 개수 중심
- 정확히 몇 개로 나눌지 정할 때는 `linspace`가 편함

```python
# arange는 step을 직접 지정
print(np.arange(0, 10, 2))

# linspace는 생성할 개수를 지정
print(np.linspace(0, 10, 6))
```

## 3.3 zeros(), ones(), empty()

- `zeros()` : 0으로 채운 배열
- `ones()` : 1로 채운 배열
- `eye()` : 단위행렬
- `full()` : 지정한 값으로 채운 배열
- `empty()` : 초기화하지 않은 배열
- `zeros_like()`, `ones_like()`, `full_like()` : 기존 배열과 같은 모양으로 생성

### np.zeros()

```python
# 0.0으로 채운 1차원 배열
a = np.zeros(5)
a
```

```python
# zeros의 기본 dtype은 float
a.dtype
```

```python
# dtype을 int로 지정
np.zeros((5,), dtype=int)
```

```python
# 2행 1열 배열 생성
b = np.zeros((2, 1))
b
```

### np.ones()

```python
# 1.0으로 채운 2행 4열 배열
np.ones((2, 4))
```

```python
# 정수형 1로 채운 배열
np.ones((2, 4), dtype=int)
```

### np.eye()

- 단위행렬 생성
- k를 사용하면 대각선 위치 조정 가능

```python
# 3x3 단위행렬
np.eye(3)
```

```python
# 정수형 단위행렬
np.eye(3, dtype=int)
```

```python
# 3행 4열 형태의 단위행렬
np.eye(3, 4)
```

```python
# k=1: 대각선이 오른쪽 위로 이동
np.eye(3, 4, 1)
```

```python
# k=-1: 대각선이 왼쪽 아래로 이동
np.eye(3, 4, -1)
```

### np.full()

- 지정한 모양을 지정한 값으로 채움
- shape과 fill_value의 모양이 맞지 않으면 오류 발생

```python
# 2행 3열을 10으로 채움
np.full([2, 3], 10)
```

```python
# fill_value가 (1, 2)이면 행마다 반복됨
np.full([2, 3], (1, 2, 3))
```

```python
# range(4)를 3행 4열에 반복
np.full([3, 4], range(4))
```

```python
# reshape을 이용해 열 방향으로 값 채우기
np.full([4, 3], fill_value=np.arange(4).reshape(4, 1))
```

```python
# 오류 예시
# (1, 2)는 길이가 2인데 shape은 (2, 3)이라 맞지 않음

np.full([2, 3], (1, 2))
```

### np.empty()

- 지정한 모양의 배열 생성
- 값을 초기화하지 않음
- 실행할 때마다 값이 다르게 보일 수 있음

```python
# 초기화하지 않은 배열
np.empty(3)
```

```python
# 2행 2열 빈 배열
np.empty([2, 2])
```

```python
# 정수형 빈 배열
np.empty([2, 3], dtype=int)
```

## 3.4 배열 연산

- numpy 배열은 원소별 연산이 가능
- 배열과 스칼라 연산은 모든 원소에 적용됨
- 같은 모양의 배열끼리는 원소별 연산 가능
- 행렬 곱은 `@`, `dot()`, `matmul()` 사용

### 배열과 스칼라 연산

```python
# 1차원 배열 생성
a = np.arange(3)
a
```

```python
# 배열의 모든 원소에 같은 연산 적용
print(a + 2)
print(a - 2)
print(a * 2)
print(a / 2)
```

### 배열과 배열의 연산

```python
# 2행 3열 배열
a = np.arange(1, 7).reshape(2, 3)
a
```

```python
# a와 같은 모양으로 배열 생성
b = np.full_like(a, [1, 2, 3])
b
```

```python
# 같은 위치의 원소끼리 연산
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

### 복합 대입 연산자

- `+=`, `-=`, `*=`, `/=` 사용 가능
- 정수 배열에 실수를 더해서 저장할 때는 dtype 때문에 오류가 날 수 있음

```python
# 정수 배열
a = np.arange(4)
a
```

```python
# 실수 배열
b = np.arange(0.5, 4, 1)
b
```

```python
# 오류 예시
# int 배열 a에 float 결과를 바로 저장하려고 해서 오류 발생 가능
a += b
```

```python
# 실수 배열 b에는 정수 배열 a를 더해 저장 가능
b += np.arange(4)
b
```

### 2차원 배열의 곱 연산자 @

- 앞 배열의 열 수와 뒤 배열의 행 수가 같아야 함
- `(2, 3) @ (3, 2)` 결과는 `(2, 2)`
- 행렬 곱은 교환법칙이 성립하지 않음

```python
# 2행 3열 배열
a = np.full((2, 3), [1, 2, 3])
a
```

```python
# 3행 2열 배열
b = np.full((3, 2), [2, 1])
b
```

```python
# @ 연산자
a @ b
```

```python
# dot() 사용
a.dot(b)
```

```python
# 순서를 바꾸면 결과 모양이 달라짐
b.dot(a)
```

```python
# matmul() 사용
print(np.matmul(a, b))
print(np.matmul(b, a))
```

### 배열 함수 sum()

```python
# 합계 계산용 배열
a = np.array([[10, 20], [5, 8], [1, 2]])
a
```

### 슬라이싱 위치에 대입

- 슬라이싱 결과에 값을 대입 가능
- 여러 값을 넣을 때는 개수가 맞아야 함
- 스칼라 하나는 선택된 부분 전체에 반복 적용됨

```python
# 배열 다시 생성
a = np.power(np.arange(10), 2)
a
```

```python
# 2~4번 위치에 -100 대입
a[2:5] = -100
a
```

```python
# 배열 다시 생성
a = np.power(np.arange(10), 2)

# 선택한 원소 3개에 값 3개 대입
a[2:5] = [-5, -5, -5]
a
```

```python
# 오류 예시
# 선택한 위치는 3개인데 대입할 값은 2개라 오류 발생
a = np.power(np.arange(10), 2)

a[2:5] = [-5, -5]
```

```python
# 스칼라 값은 선택한 위치 전체에 반복 대입됨
a = np.power(np.arange(10), 2)
a[2:5] = -10
a
```

### swapaxes(), view, copy

- `swapaxes()` : 두 축만 서로 교환
- view : 원본 데이터를 함께 참조
- copy : 원본과 별도의 새 배열 생성
- view를 수정하면 원본에도 영향을 줄 수 있음

```python
# swapaxes 확인용 배열
x = np.arange(24).reshape(2, 3, 4)
x
```

```python
# axis 0과 axis 2 교환
np.swapaxes(x, 0, 2)
```

```python
# swapaxes(x, 0, 2)는 transpose(x, (2, 1, 0))와 같은 모양
print(np.swapaxes(x, 0, 2).shape)
print(np.transpose(x, (2, 1, 0)).shape)
```

```python
# view 예시
original = np.arange(6)
view_arr = original.reshape(2, 3)

print("original:", original)
print("view_arr:")
print(view_arr)
print("view_arr.base:")
print(view_arr.base)
```

```python
# view를 수정하면 원본도 바뀜
view_arr[0, 0] = 100

print("view_arr:")
print(view_arr)
print("original:", original)
```

```python
# copy는 원본과 분리됨
original = np.arange(6)
copy_arr = original.copy()

copy_arr[0] = 100

print("original:", original)
print("copy_arr:", copy_arr)
print("copy_arr.base:", copy_arr.base)
```
