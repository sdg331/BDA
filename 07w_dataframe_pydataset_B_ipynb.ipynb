## 8.2 외부 파일 활용

- pandas는 다양한 파일 형식 읽기/쓰기 가능
- Excel 파일: read_excel(), to_excel()
- CSV 파일: read_csv(), to_csv()

### 유명한 타이타닉 파일 읽기

- titanic.xlsx 파일 읽기
- 첫 번째 열을 index로 지정

```python
# Excel 파일 읽기
import pandas as pd

titanic = pd.read_excel("titanic.xlsx", index_col=0)
titanic
```

```python
# titanic 데이터 정보 확인
titanic.info()
```

```python
# 숫자형 열 중심 통계 확인
titanic.describe()
```

```python
# 모든 열 통계 확인
titanic.describe(include='all')
```

### CSV 파일로 쓰기와 읽기

- to_csv()로 데이터프레임 저장
- read_csv()로 다시 읽기

```python
# titanic 데이터프레임을 CSV 파일로 저장
titanic.to_csv("my_titanic.csv")
```

```python
# 저장한 CSV 파일 다시 읽기
tt = pd.read_csv("my_titanic.csv", index_col=0)
tt.head(8)
```

### 인터넷 자료 파일 읽기와 쓰기

- 2022년도시도별교통사고.csv 읽기
- 인코딩은 cp949 사용

```python
import pandas as pd

# CSV 파일 읽기
traffic = pd.read_csv("/content/2022년도시도별교통사고.csv", encoding='cp949')
traffic
```

```python
# 첫 줄을 건너뛰고 첫 열을 인덱스로 지정
traffic = pd.read_csv("/content/2022년도시도별교통사고.csv",
                      encoding='cp949',
                      skiprows=1,
                      index_col=0)
traffic
```

```python
# 문자열로 읽힌 상태 확인
traffic.info()
```

```python
# 문자열 상태의 describe 확인
traffic.describe()
```

### 정수로 읽기

- thousands=',' 옵션 사용
- 콤마가 붙은 수를 정수로 읽기

```python
# 콤마가 있는 숫자를 정수로 읽기
traffic = pd.read_csv("/content/2022년도시도별교통사고.csv",
                      encoding='cp949',
                      skiprows=1,
                      index_col=0,
                      thousands=',')
traffic
```

```python
# 정수로 읽혔는지 정보 확인
traffic.info()
```

```python
# 정수형 데이터 통계 확인
traffic.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import koreanize_matplotlib

# matplotlib의 폰트를 NanumGothic으로 명시적 설정
plt.rc('font', family='NanumGothic')

# seaborn의 테마와 폰트 설정 (폰트 설정이 겹치지 않도록 주의)
sns.set_theme(font='NanumGothic', style='whitegrid')

# 마이너스 기호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False

print("한글 폰트(NanumGothic) 설정이 완료되었습니다.")
```

```python
# 두 번째 열인 사망자수(명)를 시도별 막대그래프로 표시
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(traffic[['사망자수(명)']].T)
plt.show()
```

```python
# x축 글자 회전
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(4, 3))
sns.barplot(traffic[[traffic.columns[1]]].T)
plt.xticks(rotation=90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import koreanize_matplotlib

# 한글 폰트 강제 설정
plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False

# 그래프 시각화
plt.figure(figsize=(12, 6))
sns.barplot(data=traffic[[traffic.columns[2]]].T)
plt.title('시도별 부상자수(명) 확인')
plt.xticks(rotation=45)
plt.show()
```
