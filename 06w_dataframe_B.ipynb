```python
# 여러 데이터프레임을 옆으로 비교 표시
from IPython.display import display_html

def display_side_by_side(*args):
    html_str = ''
    for df in args:
        html_str += df.to_html() + '&nbsp;' * 4
    display_html(html_str.replace('table', 'table style="display:inline"'), raw=True)
```

```python
# 원본과 행 방향 누적합 비교
display_side_by_side(df, df.cumsum(axis=1))
```

```python
# 원본과 열 방향 누적합 비교
display_side_by_side(df, df.cumsum(axis=0))
```

- 조건을 만족하는 행 검색

### 함수 df.query(조건)

- 문자열 조건식으로 행 검색
- 열 이름을 바로 사용 가능

```python
# query용 데이터프레임 생성
df = pd.DataFrame({
    'A': range(1, 6),
    'B': range(6, -3, -2),
    'C': range(10, 5, -1),
    'D clmns': range(3, -2, -1),
    'E': ['python', 'java', 'python', 'go', 'python']
})
df
```

```python
# A가 B보다 큰 행
df.query('A > B')
```

```python
# 일반 불리안 인덱싱 방식
df[df.A > df.B]
```

```python
# C가 10인 행
df.query('C == 10')
```

```python
# A가 0보다 큰 행
df.query('A > 0')
```

```python
# E가 python인 행
df.query('E == "python"')
```

```python
# E가 python이 아닌 행
df.query('E != "python"')
```

```python
# 열 이름에 공백이 있으면 백쿼트 사용
df.query('A == `D clmns`')
```

```python
# 문자열 D clmns와 비교
# 원하는 열 비교 결과가 아님
df.query('A == "D clmns"')
```

```python
# 일반 방식으로 공백 포함 열 참조
df[df.A == df["D clmns"]]
```

```python
# or 조건
df.query('A > 3 | B > 8')
```

```python
# or 조건
df.query('A > 3 or B > 8')
```

```python
# and 조건
df.query('A > 2 & C < 8')
```

```python
# and 조건
df.query('A > 2 and C < 8')
```

```python
# E가 python 또는 go인 행
df.query("E in ['python', 'go']")
```

```python
# @변수명 사용
num = 3
df.query('B > @num')
```

```python
# f 문자열 사용
num = 3
df.query(f'B > {num}')
```

```python
# 문자열 포함 검색
s = "E.str.contains('py')"
df.query(s)
```

```python
# g가 포함된 행
s = "E.str.contains('g')"
df.query(s)
```

```python
# on이 포함된 행
s = "E.str.contains('on')"
df.query(s)
```

```python
# p로 시작하는 행
s = "E.str.startswith('p')"
df.query(s)
```

```python
# n으로 끝나는 행
s = "E.str.endswith('n')"
df.query(s)
```

### fillna(value)

- dict, Series, DataFrame 사용 가능
- 리스트는 대입 불가능

```python
# dict로 열별 대체값 지정
values = {'A': 100, 'B': 200, 'C': 300, 'D': 400}
df.fillna(value=values)
```

```python
# 열별 최대 2개만 채우기
df.fillna(value=values, limit=2)
```

```python
# Series로 대체값 지정
s = pd.Series([100, 200, 300], index=['A', 'B', 'C'])
df.fillna(s)
```

```python
# DataFrame으로 대체값 지정
df_value = pd.DataFrame(np.arange(16).reshape(4, 4), columns=list('ABCD'))
df.fillna(df_value)
```

```python
# 리스트는 value로 사용할 수 없음
df.fillna([100, 200, 300, 400])
```

### 평균값으로 결측값 채우기

- 열 평균으로 결측값 대체

```python
# 열 평균 확인
df.mean()
```

```python
# 각 열의 평균으로 결측값 채우기
df.fillna(df.mean())
```

```python
# A부터 B까지의 평균만 사용
df.fillna(df.mean()['A':'B'])
```

```python
# D 열 평균 하나로 모든 결측값 채우기
df.fillna(df.mean()['D'])
```

```python
# D 열 평균으로 각 열에서 1개만 채우기
df.fillna(df.mean()['D'], limit=1)
```

```python
# axis=0 기준 limit=1
df.fillna(df.mean()['D'], limit=1, axis=0)
```

```python
# axis=1 기준 limit=1
df.fillna(df.mean()['D'], limit=1, axis=1)
```
