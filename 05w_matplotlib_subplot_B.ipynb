### 산점도

- `plt.scatter(x, y)` 사용
- `s`: 점 크기
- `c`: 색상
- `alpha`: 투명도

```python
import numpy as np
import matplotlib.pyplot as plt

# 난수 고정
np.random.seed(2025)

N = 80
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = (20 * np.random.rand(N)) ** 2

plt.scatter(x, y, s=area, c=colors, alpha=0.7)
plt.colorbar()
plt.show()
```

### 논리 인덱싱(Boolean indexing)

### 산점도와 논리 OR 조건

- `+` 연산으로 논리 OR처럼 사용
- `mask1` 또는 `mask2`를 만족하는 좌표 추출

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(-100, 100, 1000)
y = np.random.randint(-100, 100, 1000)

mask1 = abs(x) > 50
mask2 = abs(y) > 50
x = x[mask1 + mask2]
y = y[mask1 + mask2]

plt.scatter(x, y, c=x, cmap='jet', alpha=0.4)
plt.colorbar()
plt.show()
```

### 산점도와 논리 AND 조건

- `*` 연산으로 논리 AND처럼 사용
- `mask1`과 `mask2`를 모두 만족하는 좌표 추출

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(-100, 100, 1000)
y = np.random.randint(-100, 100, 1000)

mask1 = abs(x) > 50
mask2 = abs(y) > 50
x = x[mask1 * mask2]
y = y[mask1 * mask2]

plt.scatter(x, y, c=x, cmap='jet', alpha=0.5)
plt.colorbar()
plt.show()
```

```python
# figure가 두 번 생성되는 경우
# 앞의 plt.figure(figsize=(9, 4))는 사실상 반영되지 않음
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 4))
fig, axs = plt.subplots(2, 4)
# fig.set_size_inches((9, 4))
plt.show()
```

```python
# plt.subplot() 그림 제목
import matplotlib.pyplot as plt

plt.suptitle('전체 제목', fontsize=14, y=1.02)

plt.subplot(2, 3, 1)
plt.plot([1, 4, 9, 16])
plt.title('제곱수')

plt.subplot(2, 3, 6)
plt.plot([1, 2, 3, 4])
plt.title('선형')

plt.subplot(2, 3, 2)
plt.plot([6, 5, 2, 1])
plt.title('감소')

plt.tight_layout()
plt.show()
```

```python
# plt.subplots() 그림 제목
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 3, figsize=(7, 4))
fig.set_size_inches(6, 3)

ax[0, 1].plot([1, 4, 9, 16])
ax[1, 0].plot([1, 2, 3, 4])
ax[1, 2].plot([6, 5, 2, 1])

fig.suptitle('전체 제목', fontsize=14, fontweight='bold')

ax[0, 0].set_title('(0,0) 빈 플롯')
ax[0, 1].set_title('(0,1) 제곱수')
ax[0, 2].set_title('(0,2) 빈 플롯')
ax[1, 0].set_title('(1,0) 선형')
ax[1, 1].set_title('(1,1) 빈 플롯')
ax[1, 2].set_title('(1,2) 감소')

plt.tight_layout()
plt.show()
```

```python
# fig.add_subplot() 그림 제목
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(6, 4))

ax1 = fig.add_subplot(231)
ax1.plot([1, 4, 9, 16])
ax1.set_title('제곱수')

ax2 = fig.add_subplot(233)
ax2.plot([1, 2, 3, 4])
ax2.set_title('선형')

ax3 = fig.add_subplot(235)
ax3.plot([6, 5, 2, 1])
ax3.set_title('감소')

fig.suptitle('전체 제목', fontsize=14)

plt.tight_layout()
plt.show()
```

### plt.subplots()과 fig.add_subplot() 비교

| 구분 | plt.subplots() | fig.add_subplot() |
|---|---|---|
| figure 생성 | 자동 생성 | plt.figure()로 먼저 생성 |
| Axes 생성 | 한 번에 여러 개 생성 | 하나씩 추가 |
| 코드 | 간단함 | 세밀한 제어 가능 |
| 사용 상황 | 다수의 서브플롯 | 복잡한 레이아웃 |

### 시리즈 논리 참조

- 조건 결과가 True인 값만 추출
- `&`, `|`, `~` 연산 사용 가능

```python
import pandas as pd

s = pd.Series([3, 6, -3, 0, -4, 8, -7])
s
```

```python
# 0보다 큰지 확인
s > 0
```

```python
# 짝수인지 확인
s % 2 == 0
```

```python
# 0보다 큰 값만 추출
s[s > 0]
```

```python
# 0보다 크지 않은 값 추출
s[~(s > 0)]
```

```python
# 짝수만 추출
s[s % 2 == 0]
```

```python
# 홀수만 추출
s[s % 2 != 0]
```

```python
# 0보다 크고 짝수인 값 추출
s[(s > 0) & (s % 2 == 0)]
```

```python
# 조건 결과도 시리즈
n = s > 0
n
```

```python
# 조건 결과 자료형 확인
type(n)
```

### count(), value_counts()

- NaN, None은 결측값
- `count()`는 결측값 제외
- `value_counts()`는 값의 빈도 확인

```python
import numpy as np
import pandas as pd

s = pd.Series([3, 1, 1, 2, None, np.nan, 4, 6, 6, np.nan])
s
```

```python
# 실제 값 확인
s.values
```

```python
# 결측값 제외 원소 수
s.count()
```

```python
# 값의 빈도 확인
s.value_counts()
```

```python
# 결측값 포함 빈도 확인
s.value_counts(dropna=False)
```

```python
# 정렬하지 않고 결측값 포함
s.value_counts(sort=False, dropna=False)
```

```python
# 구간을 4개로 나누어 빈도 확인
s.value_counts(bins=4)
```

```python
# 구간 빈도, 정렬하지 않음
s.value_counts(bins=4, sort=False)
```

```python
# 직접 구간 지정
s.value_counts(bins=[.5, 3.5, 6.5], sort=False)
```

## 6.3 DataFrame의 다양한 참조

- 열 참조
- 조건에 맞는 행 참조
- loc, iloc, at, iat

```python
# 참조 실습용 데이터프레임
import pandas as pd

pf = pd.DataFrame(data=[[25, 35, 8, 18],
                        [18, 27, 10, 20],
                        [17, 17, 10, 19],
                        [12, 22, 9, 20],
                        [22, 34, 8, 16]],
                  index=['윤일형', '강수희', '홍소희', '유한빈', '신수빈'],
                  columns=['중간', '기말', '과제', '출석'])
pf
```

### 데이터프레임 열 참조

- `df['열명']`: Series 반환
- `df.열명`: Series 반환
- `df[['열명']]`: DataFrame 반환

```python
# 열 하나를 Series로 참조
pf['중간']
```

```python
# 열 하나를 속성처럼 참조
pf.중간
```

```python
# 열 하나를 DataFrame으로 참조
pf[['중간']]
```

```python
# 여러 열 참조
pf[['중간', '기말']]
```

### 조건에 맞는 행 참조

- 조건식 결과가 True인 행만 선택

```python
# 중간 점수가 20 이상인 행
pf[pf['중간'] >= 20]
```

```python
# 과제가 9 이상인 행
pf[pf['과제'] >= 9]
```

```python
# 중간 20 이상이고 출석 18 이상인 행
pf[(pf['중간'] >= 20) & (pf['출석'] >= 18)]
```

### df.loc[]

- 레이블 기반 행과 열 참조
- 행 이름, 열 이름을 사용

```python
# 첫 번째 행 이름 확인
pf.index[0]
```

```python
# 첫 번째 행을 loc로 참조
pf.loc[pf.index[0]]
```

```python
# 행 이름으로 한 행 참조
pf.loc['윤일형']
```

```python
# 여러 행 이름으로 참조
pf.loc[['윤일형', '홍소희']]
```

```python
# 행 이름 슬라이싱
# loc 슬라이싱은 끝 레이블 포함
pf.loc['윤일형':'홍소희']
```

```python
# 조건을 loc에 사용
pf.loc[pf['중간'] >= 20]
```

### loc로 열 참조

- `df.loc[:, '열명']`
- `df.loc[:, df.columns[0]]`

```python
# 중간 열 전체 참조
pf.loc[:, '중간']
```

```python
# 첫 번째 열 이름 확인
pf.columns[0]
```

```python
# 첫 번째 열을 loc로 참조
pf.loc[:, pf.columns[0]]
```

```python
# 여러 열 참조
pf.loc[:, ['중간', '과제']]
```

```python
# 행과 열을 함께 참조
pf.loc['윤일형', '중간']
```

```python
# 행과 열 범위 참조
pf.loc['윤일형':'홍소희', '중간':'과제']
```

### df.iloc[]

- 정수 위치 기반 참조
- 0부터 시작하는 위치 번호 사용

```python
# 첫 번째 행 참조
pf.iloc[0]
```

```python
# 여러 행 참조
pf.iloc[[0, 2, 4]]
```

```python
# 행 슬라이싱
pf.iloc[1:4]
```

```python
# 첫 번째 열 참조
pf.iloc[:, 0]
```

```python
# 여러 열 참조
pf.iloc[:, [0, 2]]
```

```python
# 행과 열 위치로 하나의 값 참조
pf.iloc[0, 0]
```

```python
# 행과 열 위치 범위 참조
pf.iloc[0:3, 0:3]
```

### at, iat

- `at`: 레이블 기반 단일 값 참조
- `iat`: 정수 위치 기반 단일 값 참조

```python
# at으로 단일 값 참조
pf.at['윤일형', '중간']
```

```python
# iat으로 단일 값 참조
pf.iat[0, 0]
```

```python
# at으로 값 수정
pf.at['윤일형', '중간'] = 26
pf
```

```python
# iat으로 값 수정
pf.iat[0, 0] = 25
pf
```
