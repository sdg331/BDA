```python
pop.columns = ['총인구수', '세대수', '세대당인구', '남자인구수', '여자인구수', '남여비율']
pop.head() # 첫 5행을 표시
```

```python
# 코랩이 추천하는 그래프
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 7))
pop[['남자인구수', '여자인구수']].plot(kind='bar', figsize=(12, 7))
plt.title('행정구역별 남녀 인구수 비교')
plt.xlabel('행정구역')
plt.ylabel('인구수')
plt.xticks(rotation=90)
plt.legend(['남자인구수', '여자인구수'])
plt.tight_layout()
plt.show()
```

```python
# 인구 데이터 시각화
# seaborn의 sns.scatterplot()으로 '총인구수'의 산점도를 그린 코드
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(5, 3))
sns.scatterplot(pop['총인구수'])
plt.xticks(rotation=90)
plt.show()
```

```python
pop.sort_values('총인구수')
```

```python
# 총 인구수로 정렬하여 막대 그래프를 그린 코드
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(5, 3))

pop1 = pop.sort_values('총인구수')
sns.barplot(x=pop1.index, y='총인구수', data=pop1)
plt.xticks(rotation=90)
plt.show()
```
